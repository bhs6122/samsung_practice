```python
# ■ 연습문제
# ● 내용을 저장하고 저장된 내용을 읽을 수 있는 Letter 클래스 작성하기

class Letter:
# 생성자 작성, 변수 초기화
# write 메소드 작성
# show 메소드 작성
letter1 = Letter()
letter1.write('가나다라')
letter1.write('카타파하')
print(letter1.show())
letter2 = Letter()
letter2.write('ABCD')
letter2.write('WXYZ')
print(letter2.show())
```

```python
#위 연습문제 답
class Letter:
    def __init__(self):
        self.data = ''
    
    def write(self, data):
        self.data += data

    def show(self):
        return self.data

letter1 = Letter()
letter1.write('가나다라')
letter1.write('카타파하')
print(letter1.show())
letter2 = Letter()
letter2.write('ABCD')
letter2.write('WXYZ')
print(letter2.show())
```

```python
# 위 연습문제 답

class Car:
    def __init__(self):
        self.opt = []

    def add_option(self, opt):
        self.opt.append(opt)        #self.opt가 리스트이므로 append사용

    def show_option(self):
        return self.opt

car1 = Car()    # Car의 인스턴스: 클래스를 메모리에 객체로 생성
car2 = Car()

car1.add_option('전동 트렁크')
car1.add_option('통풍 시트')
print(car1.show_option())

car2.add_option('뒷자리 에어백')
car2.add_option('하이패스')
print(car2.show_option())
```

```python
# ■ 연습문제
# ● 데이터베이스를 활용하는 코드에 예외처리 적용하기
# 1. SQL 표현이 잘못된 경우 예외처리
# 2. 어떤 경우로 종료가 되더라도 close()가 수행되도록 finally 처리

import sqlite3
conn = sqlite3.connect('geoje.db')
cursor = conn.cursor()

sql = '''
SELECT id, name, desc, mapinfo, latitude, longitude
FORM tour
''' # FROM tour (일부러 오타내서 오류상황 발생)

try:
    cursor.execute(sql)
    result = cursor.fetchmany(10)
    for row in result:
        print(row[1], row[4], row[5])

except:
    print('?? 오류 발생, 어떻게 처리해라..')

finally:
    cursor.close()
    conn.close()
```

```python
# ■연습문제
# ●쇼핑몰 정보 및 상품 현황 출력하기
# - 요청 URL: http://ggoreb.com/python/api/request_market.jsp
# - 요구사항
# 1) 응답 데이터 구조 확인 후 "쇼핑몰 이름"과 "전체 상품개수" 출력
# 2) 모든 상품명과 가격 출력

url = 'http://ggoreb.com/python/api/request_market.jsp'
res = requests.get(url)

#print(res.text)
data = res.json()
#print(type(data))

print(f'[{data['market_name']} 상품 현황 - 총 {data['total_count']}개]')


for i in data["items"]:
    print(f'- 상품명: {i['title']} | 가격: {i['price']}')
```

```python
# 연습문제
# ●전국 실시간 미세먼지 및 대기질 현황출력하기
# - 요청 URL: http://ggoreb.com/python/api/request_weather.jsp
# - 요구사항
# 1) 전체 지역 중 대기질이 나쁨인 지역만 필터링 후 출력
# (state 나쁨 또는 action_required True)
# 2) 출력 형식

url = 'http://ggoreb.com/python/api/request_weather.jsp'
res = requests.get(url)

#print(res.text)
data = res.json()

print(f'[{data['data_source']} - {data['update_time']} 기준 위험 지역 안내]')

for i in data['regional_air_quality']:
    if i['status']['state'] == "나쁨":
        print(f'경보 지역: {i['location']} | 미세먼지 농도: {i['pm10']} (상태: {i['status']['state']})')
```
