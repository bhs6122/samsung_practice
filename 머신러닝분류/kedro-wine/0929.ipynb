```python
from kedro.pipeline import node
def add(a, b):
    return a + b
n = node(add, inputs=["left", "right"],
outputs="total", name="add_numbers")
print(n.run({"left": 2, "right": 3}))
# {"total": 5}
```

```python
# 순서대로 함수 인수에 전달
n1 = node(add, ["left", "right"], "total")

# 함수 인수 이름을 명시해 전달
n2 = node(add, {"a": "left", "b": "right"}, "total")
print(n2.run({"left": 2, "right": 3}))
# {"total": 5}
```

```python
def extremes(values):
    return min(values), max(values)
n = node(extremes, "values", ["minimum", "maximum"], name="find_extremes")
print(n.run({"values": [2, 5, 9]}))
# {"minimum": 2, "maximum": 9}
```
