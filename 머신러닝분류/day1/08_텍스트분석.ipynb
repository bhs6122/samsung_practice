2. 텍스트 토큰화: 문장을 분리하고, 단어를 분리하여 토큰으로 만드는 작업

```python
# 문장 토큰화
# 토큰화(tokenize): 텍스트를 의미있는 조각(토큰)으로 쪼개는 작업
import nltk
    # - nlkt: 영어 텍스트 처리용 대표 라이브러리(Natural Language Toolkit)

nltk.download('punkt_tab')
    # - punk_tab: 문장 경계를 판단하는 학습된 모델
    # - 라이브러리 본체와 별개로, 작업에 필요한 것들을 download()로 따로 받아야함
    # - 얘를 다운받아야 문장이나 단어를 분리하고 토큰화 할 수 있음
    # - 마침표가 문장 끝인지, Mr./ U.S. 같은 약어인지 구분해줌
```

```python
sentences = nltk.sent_tokenize(text = text)
    # - sent_tokenize: 긴 글 하나를 문장 단위로 잘라 리스트로 만듬
    #   (punkt_tab을 다운받아야 sent_tokenize명령어를 쓸 수 있음)
    # - word_tokenize: 단어 단위로 쪼갬
    # - ★ 앞 셀에서 정규식으로 클렌징할 때 마침표(\.)를 일부러 남긴 이유가 이것

print(sentences[:3])
    # - 앞 3문장만 확인(전체로 찍으면 너무 김)
```

```python
# sent_tokenize 결과 간단 예시
example_sentences = nltk.sent_tokenize(text = 'im hungry. i want to eat')

print(example_sentences)

    # 결과를 보면 마침표 기준으로 문장이 나눠진 것을 확인 할 수 있음
```

```python
# 단어 토큰화
word_token = []
for sentence in sentences:
    words = nltk.word_tokenize(sentence)
    word_token.extend(words)
print(word_token[:10])
```

```python
import pandas as pd

review_df = pd.read_csv('./data/labeledTrainData.tsv', header=0, sep="\t", quoting=3)
review_df.head(3)
```

```python
from sklearn.model_selection import train_test_split 

class_df = review_df['sentiment']
feature_df = review_df.drop(['id','sentiment'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(feature_df, class_df, test_size=0.2, random_state=156)
print(X_train.shape, X_test.shape)
```

- #### 유사도 분석
  - 코사인 유사도 실습

```python
import numpy as np

def cos_similarity(v1, v2):
    dot_product = np.dot(v1, v2)
    l2_norm = (np.sqrt(sum(np.square(v1))) * np.sqrt(sum(np.square(v2))))
    similarity = dot_product / l2_norm

    return similarity
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

doc_list = ['if you take the blue pill, the story ends',
            'if you take the red pill, you stay in Wonderland',
            'if you take the red pill, I show you how deep the rabbit hole goes']

tfidf_vect_simple = TfidfVectorizer()
feature_vect_simple = tfidf_vect_simple.fit_transform(doc_list)
print(feature_vect_simple.shape)
```

```python
feature_vect_array = feature_vect_simple.toarray()

vect1 = np.array(feature_vect_array[0]).reshape(-1, )
vect2 = np.array(feature_vect_array[1]).reshape(-1, )

similarity_simple = cos_similarity(vect1, vect2)
print('문장1, 문장2 코사인유사도: {0:.3f}'.format(similarity_simple))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_simple_pair = cosine_similarity(feature_vect_simple, feature_vect_simple)

print(similarity_simple_pair)
print('shape:',similarity_simple_pair.shape)
```
