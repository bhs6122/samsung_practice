- 텍스트 전처리
  1. 클렌징: 텍스트에서 불필요한 문자, 기호 등을 사전에 제거한다.

```python
with open('./data/stevejobs.txt', 'r', encoding='utf8') as f:
    rows = f.readlines()
    lines = [row for row in rows]
text = ' '.join(lines)
print(text[:200])
```

```python
import re

compile = re.compile("[^ a-zA-Z0-9\\.]+")
text = compile.sub('',text).lower()
print(text[:200])
```

2. 텍스트 토큰화: 문장을 분리하고, 단어를 분리하여 토큰으로 만드는 작업

```python
# 문장 토큰화
# 토큰화(tokenize): 텍스트를 의미있는 조각(토큰)으로 쪼개는 작업
import nltk
    # - nlkt: 영어 텍스트 처리용 대표 라이브러리(Natural Language Toolkit)

nltk.download('punkt_tab')
    # - punk_tab: 문장 경계를 판단하는 학습된 모델
    # - 라이브러리 본체와 별개로, 작업에 필요한 것들을 download()로 따로 받아야함
    # - 얘를 다운받아야 문장이나 단어를 분리하고 토큰화 할 수 있음
    # - 마침표가 문장 끝인지, Mr./ U.S. 같은 약어인지 구분해줌
```

```python
sentences = nltk.sent_tokenize(text = text)
    # - sent_tokenize: 긴 글 하나를 문장 단위로 잘라 리스트로 만듬
    #   (punkt_tab을 다운받아야 sent_tokenize명령어를 쓸 수 있음)
    # - word_tokenize: 단어 단위로 쪼갬
    # - ★ 앞 셀에서 정규식으로 클렌징할 때 마침표(\.)를 일부러 남긴 이유가 이것

print(sentences[:3])
    # - 앞 3문장만 확인(전체로 찍으면 너무 김)
```

```python
# sent_tokenize 결과 간단 예시
example_sentences = nltk.sent_tokenize(text = 'im hungry. i want to eat')

print(example_sentences)

    # 결과를 보면 마침표 기준으로 문장이 나눠진 것을 확인 할 수 있음
```

```python
# 단어 토큰화
word_token = []
for sentence in sentences:
    words = nltk.word_tokenize(sentence)
    word_token.extend(words)
print(word_token[:10])
```

```python
nltk.word_tokenize(text = 'imhungrydoyou')
    # 띄어쓰기 안하면 안나눠짐
```

```python
nltk.word_tokenize(text = '나는 배가 고프다')

# 한글은 nltk잘 안쓴다고 하는데
```

3. 스톱워드 제거: 분석에 큰 의미 없는 단어 제거(this, is 등)

```python
nltk.download('stopwords')
```

```python
stopwords = nltk.corpus.stopwords.words('english')
print('영어 stop words 개수: ',len(stopwords))
print(stopwords[:10])
```

```python
filtered_words = []
for word in word_token:
    if word not in stopwords:
        filtered_words.append(word)
print(filtered_words[:10])
```

- Stemming과 Lemmatization: 단어의 원형을 찾는 기법

```python
from nltk.stem import LancasterStemmer
stemmer = LancasterStemmer()

print(stemmer.stem('working'))
print(stemmer.stem('works'))
print(stemmer.stem('worked'))
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemma = WordNetLemmatizer()

print(lemma.lemmatize('amusing','v'))
print(lemma.lemmatize('amuses','v'))
print(lemma.lemmatize('amused','v'))
```

```python
print(stemmer.stem('happier'), stemmer.stem('happiest'))
print(lemma.lemmatize('happier','a'))
print(lemma.lemmatize('happiest','a'))
```

```python
filtered_words = []
for word in word_token:
    stemmed_word = stemmer.stem(word)
    if stemmed_word not in stopwords:
        filtered_words.append(stemmed_word)
print(filtered_words[:10])

# stemmed를 사용하면 this -> thi, pixar -> pix, apple -> appl로 이상하게 잘린다.
# 이상하게 잘리기 때문에 요즘에는 stemming은 잘 사용하지 않는다.
```

- 텍스트 분류 실습 - 20 뉴스 그룹

```python
from sklearn.datasets import fetch_20newsgroups
import pandas as pd

news_data = fetch_20newsgroups(subset = 'all', random_state = 156)
print(news_data.data[0])
```

```python
news_data.target_names

# 이 문장이 어떤 카테고리에 속해있냐 이거ㅣㅇㅁ

# comp는 computer를 의미
# comp.graphics는 컴퓨터에서 그래픽 관련 내용 뭐 이런식
```

```python
train_news= fetch_20newsgroups(subset='train', 
                               remove=('headers', 'footers', 'quotes'), 
                               random_state=156
                               )
X_train = train_news.data
y_train = train_news.target
print(X_train[0])

# 이 결과가 하나의 텍스트 데이터(문서)
```

```python
test_news= fetch_20newsgroups(subset='test', 
                              remove=('headers', 'footers','quotes'), 
                              random_state=156
                              )
X_test = test_news.data
y_test = test_news.target
print(f'학습 데이터 크기 {len(train_news.data)}, 테스트 데이터 크기 {len(test_news.data)}')
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cnt_vect = CountVectorizer()
cnt_vect.fit(X_train)
X_train_cnt_vect = cnt_vect.transform(X_train)
X_test_cnt_vect = cnt_vect.transform(X_test)

print('학습 데이터 Text의 CountVectorizer Shape:', X_train_cnt_vect.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cnt_vect = CountVectorizer()
cnt_vect.fit(X_train)
X_train_cnt_vect = cnt_vect.transform(X_train)
X_test_cnt_vect = cnt_vect.transform(X_test)

print('학습 데이터 Text의 CountVectorizer Shape:', X_train_cnt_vect.shape)
```

```python
X_train_cnt_vect
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import warnings
warnings.filterwarnings('ignore')

lr_clf = LogisticRegression()
lr_clf.fit(X_train_cnt_vect , y_train)
pred = lr_clf.predict(X_test_cnt_vect)
print('예측 정확도는 {0:.3f}'.format(accuracy_score(y_test, pred)))
```
