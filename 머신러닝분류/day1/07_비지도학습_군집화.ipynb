### K-평균 군집화(K-means clustering)
 - 아이리스 데이터셋을 사용한 k-평균 군집화 실습

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

iris = load_iris()
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
irisDF = pd.DataFrame(data=iris.data, columns = columns)
```

```python
kmeans = KMeans(n_clusters=3, max_iter=300, random_state=0)
    # [반복]: K-means가 하는일
    # 1. 중심점 3개를 랜덤하게 찍음
    # 2. 각 iris를 가장 가까운 중심점에 배정(소속)시킴
    # 3. 배정된 iris들의 평균 위치로 중심점을 옮김
    # 4. 중심점이 더 안 움직일 때까지 반복

    # - 거리로 묶기 때문에 단위가 큰 열이 결과에 크게 영향을 미침
    #   → 그래서 보통 StandardScaler로 스케일을 맞춘 뒤에 넣음
    #     (현재 데이터셋인 iris는 4개 열이 전부 cm에 값 범위도 비슷해서 안했음)

    # - n_clusters: 몇덩어리로 묶을지 사람이 지정함(k)
    #               알고리즘이 스스로 못 정하는 값임.
    #               이 데이터셋의 iris 품종이 3종인걸 미리 알고 있기 때문에 3으로 설정한거.

    # - max_iter: [반복] 과정을 최대 몇 번까지 돌릴지(기본값 300)
    #             보통 그 전에 중심점이 안움직여서 알아서 멈추긴함

    # - random_state: 맨 처음 중심점 3개를 랜덤하게 찍기 때문에, 고정해야 매번 같은결과가 나옴

    # 


kmeans.fit(irisDF)
```

```python
print(kmeans.labels_)

# 여기서 1은 그냥 1집단을 의미 숫자는 임의로 붙인거.
# 원래 데이터의 1과는 관계없음
```

```python
# 원본 데이터
print(iris.target)

    # 
```

```python
irisDF['target'] = iris.target
irisDF['cluster']=kmeans.labels_
iris_result = irisDF.groupby(['target', 'cluster'])['sepal_length'].count()
print(iris_result)
```

```python
from sklearn.decomposition import PCA

scaler = StandardScaler()
scaled = scaler.fit_transform(irisDF.iloc[:,:4])
pca = PCA(n_components=2)
pca_transformed = pca.fit_transform(scaled)

irisDF['PC1'] = pca_transformed[:,0]
irisDF['PC2'] = pca_transformed[:,1]
```

```python
markers = ['^', 's', 'o']
for i, marker in enumerate(markers):
    x_axis_data = irisDF[irisDF['cluster'] == i]['PC1']
    y_axis_data = irisDF[irisDF['cluster'] == i]['PC2']
    plt.scatter(x_axis_data, y_axis_data, marker = marker, label = iris.target_names[i])
plt.legend()
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

```python
# 최적의 군집 개수 찾기

inertia = []
for i in range(1,5):
    kmeans_plus = KMeans(n_clusters = i, max_iter=300, random_state=0)
    kmeans_plus.fit(irisDF.iloc[:,:4])
    inertia.append(kmeans_plus.inertia_)

plt.plot(range(1,5), inertia, 'o-')
plt.xlabel('Number of Clusters')
plt.ylabel('Inertia')
plt.show()
```

### 군집 평가(Cluster Evaluation)
 - 실루엣 분석을 사용한 군집 평가 실습

```python
from sklearn.metrics import silhouette_samples, silhouette_score

score_samples = silhouette_samples(iris.data, irisDF['cluster'])
print('silhouette_samples( ) return 값의 shape' , score_samples.shape)

irisDF['silhouette_coeff'] = score_samples
average_score = silhouette_score(iris.data, irisDF['cluster'])
print('붓꽃 데이터셋 Silhouette Analysis Score:{0:.3f}'.format(average_score))
irisDF.head()
```

```python
irisDF.groupby('cluster')['silhouette_coeff'].mean()
```

### 평균 이동(Mean Shift)
 - 평균 이동 실습

```python
from sklearn.cluster import MeanShift

meanshift = MeanShift(bandwidth = 0.8)
cluster_labels = meanshift.fit_predict(irisDF.iloc[:, :4])
print('클러스터 개수 :', len(np.unique(cluster_labels)))
```

```python
from sklearn.cluster import estimate_bandwidth

bandwidth = estimate_bandwidth(irisDF.iloc[:, :4])
print('bandwidth 값:', round(bandwidth, 3))

meanshift = MeanShift(bandwidth = bandwidth)
cluster_labels = meanshift.fit_predict(irisDF.iloc[:, :4])
irisDF['meanshift'] = cluster_labels
print('클러스터 개수 :', len(np.unique(cluster_labels)))
```

```python
markers = ['^', 's']

for i, marker in enumerate(markers):
    x_axis_data = irisDF[irisDF['meanshift'] == i]['PC1']
    y_axis_data = irisDF[irisDF['meanshift'] == i]['PC2']
    plt.scatter(x_axis_data, y_axis_data, marker = marker)

plt.xlabel('PC1')
plt.ylabel('PC2')
plt.show()
```

### GMM(Gaussian Mixture Model)

```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components = 3, random_state = 0)
gmm.fit(irisDF.iloc[:,:4])
gmm_labels = gmm.predict(irisDF.iloc[:,:4])
irisDF['gmm'] = gmm_labels
print(gmm_labels)
```

```python
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

iris = load_iris()
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

irisDF = pd.DataFrame(data=iris.data, columns = columns)
irisDF['target'] = iris.target
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps = 0.6, min_samples = 8, metric = 'euclidean')
    #           eps: epsilon을 의미
dbscan_labels = dbscan.fit_predict(iris.data)

irisDF['dbscan_cluster'] = dbscan_labels
irisDF['target'] = iris.target

iris_result = irisDF.groupby(['target'])['dbscan_cluster'].value_counts()
print(iris_result)
```

### 이상치 탐지(Anomaly Detection)

```python
from sklearn.ensemble import IsolationForest
from sklearn.datasets import load_iris

iris = load_iris()
iso = IsolationForest(contamination=0.05, random_state = 0)

iso.fit(iris.data)
pred = iso.predict(iris.data) # 이상치는 -1, 정상치는 1
print('이상치로 판별된 데이터 개수 :', (pred == -1).sum())
```
