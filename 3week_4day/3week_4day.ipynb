```python
# 라이브러리 호출

from sklearn.datasets import load_iris  # datasets은 사이킷런의 데이터 셋을 생성하는 모듈이다.
from sklearn.tree import DecisionTreeClassifier # 트리 기반 ML 알고리즘을 구현하는 모듈이다.
from sklearn.model_selection import train_test_split    
# model_selection은 학습 데이터와 검증 데이터, 예측 데이터로 데이터를 분리하는 다양한 모듈이다.
# train_test_split 함수는 데이터를 학습, 테스트 데이터로 분리하는 함수이다.

# 대문자로 시작하면 class일 가능성이 높고, 소문자로 시작하면 함수일 가능성이 높다.
# 전부 대문자면 변수일 가능성이 높다.
```

```python
iris = load_iris()
iris_data = iris.data       # iris_data = iris['data']와 같음
iris_label = iris.target
print('iris target값:', iris_label)
print('iris target명:', iris.target_names)
iris_df = pd.DataFrame(data = iris_data, columns = iris.feature_names)
iris_df['label'] = iris.target
iris_df.head(3)
```

```python
iris['target_names'][0]
```

```python
X_train, X_test, y_train, y_test = train_test_split(iris_data, iris_label,
                                                    test_size = 0.2,
                                                    random_state = 11
                                                    )
print(X_train[0])
```

```python
# 모델 선택
dt_clf = DecisionTreeClassifier(random_state = 11)

# 학습
dt_clf.fit(X_train, y_train)        # fit(학습용_문제, 학습용_정답)

# 예측
pred = dt_clf.predict(X_test)

# 평가
from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test, pred)      # 평가한 점수를 반환
print('예측 정확도: {0: 4f}'.format(acc))

# 머신러닝 모델에다 데이터를 넣어주는 것을 학습
# fit이 학습하는거
# X_train과 y_train 모두 주고 학습. 문제와 정답을 주고 학습시키는 지도학습 방식
# X_test라는 테스트데이터 독립변수 즉 문제만 주고 결과를 예측하게함
# 이때 pred변수가 예측값

# 머신러닝은 학습 -> 예측 -> 평가 단계로 이루어짐
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import KFold

iris = load_iris()
features = iris.data
label = iris.target
dt_clf= DecisionTreeClassifier(random_state= 156)
```

```python
# 학습 단계
kfold= KFold(n_splits= 5)
cv_accuracy= []
n_iter= 0

for train_index, test_index in kfold.split(features):
    X_train, X_test= features[train_index], features[test_index]
    y_train, y_test= label[train_index], label[test_index]

    dt_clf.fit(X_train, y_train)
    pred = dt_clf.predict(X_test)
    n_iter+= 1
    accuracy = np.round(accuracy_score(y_test, pred), 4)
    train_size= X_train.shape[0]
    test_size= X_test.shape[0]

    print('#',n_iter)
    print('교차검증정확도:',accuracy)
    print('학습데이터크기:',train_size)
    print('검증데이터크기:',test_size)
    print('검증인덱스:',test_index)

    cv_accuracy.append(accuracy)
print('평균검증정확도:', np.mean(cv_accuracy)) 
```

```python
# 교차 검증
from sklearn.model_selection import StratifiedKFold
skf= StratifiedKFold(n_splits= 3)
n_iter= 0
for train_index, test_index in skf.split(iris_df, iris_df['label']):
    n_iter+= 1
    label_train= iris_df['label'].iloc[train_index]
    label_test= iris_df['label'].iloc[test_index]
    print(f'#{n_iter} 교차검증')
    print('학습레이블분포:\n', label_train.value_counts())
    print('검증레이블분포:\n', label_test.value_counts())
```

```python
# 교차 검증
from sklearn.model_selection import cross_val_score
iris_data= load_iris()
dt_clf= DecisionTreeClassifier(random_state= 156)
data = iris_data.data
label = iris_data.target        # 이전 코드에 이미 불러와서 굳이 쓸 필요는 없음
scores = cross_val_score(dt_clf, data, label, scoring = 'accuracy', cv = 5)
print('교차검증별정확도: ', np.round(scores, 4))
print('평균검증정확도: ', np.round(np.mean(scores), 4))
# 이 교차 검증 방법이 더 좋다. 
# 교차 검증을 왜 쓰는지 아는 것이 더 중요하다.
```

```python
from sklearn.model_selection import GridSearchCV

iris = load_iris()
X_train, X_test, y_train, y_test= train_test_split(
    iris_data.data, iris_data.target, 
    test_size=0.2, random_state=121
    )
dtree= DecisionTreeClassifier()
parameters = {'max_depth': [1, 2, 3], 'min_samples_split': [2, 3]}
grid_dtree = GridSearchCV(dtree, param_grid=parameters, cv=3, refit=True)
grid_dtree.fit(X_train, y_train)
```

```python
scores_df = pd.DataFrame(grid_dtree.cv_results_)
scores_df[['params', 'mean_test_score', 'rank_test_score',
           'split0_test_score', 'split1_test_score', 'split2_test_score']]
```
