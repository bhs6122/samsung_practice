```python
# 문제 3) 회원가입 폼 유효성 검사하기 (논리/검사 내장함수)
#         회원가입 시 입력받은 정보가 조건을 만족하는지 검사하는 함수를 만드세요.

def check_signup(user_id, password, age):
   items = [
      # 조건 1) user_id는 문자열이어야 함
      isinstance(user_id, str),
      # 조건 2) password 길이는 8자 이상이어야 함
      len(password) >= 8,
      # 조건 3) age는 정수형이며 14세 이상이어야 함
      isinstance(age, int) and age >= 14
   ]
   return all(items)

print(check_signup('minjun01', 'pw12345678', 20))     # True
print(check_signup('minjun01', 'pw123', 20))          # False (비밀번호 짧음)
print(check_signup('minjun01', 'pw12345678', '20'))   # False (age가 문자열)
```

```python
stock_raw = ['120', '35', '8', '250', '64']

items = list(map(lambda x: int(x), stock_raw))
#items   #결과: [120, 35, 8, 250, 64]

low_stock = list(filter(lambda x: x < 50, items))
print('재고부족 목록:', low_stock)

high_stock = list(filter(lambda x: x >= 100, items))
transfer_list = list(map(lambda x: x // 2, high_stock))
print('이관 수량:', transfer_list)
```

```python
import shutil
import smtplib
from email.mime.text import MIMEText
from email.mime.multipart import MIMEMultipart

SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 587
SENDER = "amugona.2606@gmail.com"
PASSWORD = "orzk gmnw uhmu bgqp"
RECIPIENT = "seorab@naver.com"

def check_disk_usage(path="/", threshold=80):
    total, used, free = shutil.disk_usage(path)
    usage = used / total * 100
    return usage, threshold

def send_email(subject, body):
    msg = MIMEMultipart()
    msg["From"] = SENDER
    msg["To"] = RECIPIENT
    msg["Subject"] = subject
    msg.attach(MIMEText(body, "html"))

    with smtplib.SMTP(SMTP_SERVER, SMTP_PORT) as server:
        server.starttls()
        server.login(SENDER, PASSWORD)
        server.send_message(msg)

if __name__ == "__main__":
    usage, threshold = check_disk_usage("/")
    subject = f"[경고] 디스크 사용량 {usage:.1f}% 초과"
    body = f"""
    <h3>⚠️ 디스크 경고</h3>
    <p>현재 디스크 사용량은 <b>{usage:.1f}%</b> 입니다.<br>
    서버 점검이 필요합니다.</p>
    """
    try:
        send_email(subject, body)
        print("⚠️ 이메일 발송 완료")
    except Exception as e:
        print(f"❌ 이메일 발송 실패: {e}")
```

```python
# 연습문제
# 아래의 코드가 실행되어 결과와 같이 출력될 수 있도록 메소드 작성하기

class Book:

# 코드 작성
    def setTitle(self, title):
        self.title = title
    def setPrice(self, price):
        self.price = price

book1 = Book()
book1.setTitle('Python')
book1.setPrice(10000)   # 생성 후 .을 사용해서 불리는 함수: 인스턴스 메소드
print(book1.title, book1.price)

book2 = Book()
book2.setTitle('Java')
book2.setPrice(10000)
print(book2.title, book2.price)
```

```python
# 연습문제
# 아래의 코드가 실행되어 결과와 같이 출력될 수 있도록 생성자 작성하기

class Product:
    def __init__(self, name, code, count):      #self는 그냥 써야됨
        self.name = name
        self.code = code
        self.count = count

    def show_info(self):
        return 'name=%s, code=%s, count=%s' % (self.name, self.code, self.count)

p1 = Product('clock', 'c0001', 500) # 생성하면서 값을 넣겠다는 의미
print(p1.show_info())

p2 = Product('tv', 't0001', 1500)
print(p2.show_info())
```
