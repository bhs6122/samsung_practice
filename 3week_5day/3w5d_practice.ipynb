```python
# housing_price 데이터 실습 해보기
house = pd.read_csv('data/Housing_price.csv')
house.head()
data = house.iloc[:, :5]
data.head()
# 해야할것: 인코딩, 스케일링
```

```python
# 회귀 데이터에서 price같이 너무 큰 값을 스케일링해서 숫자를 작게 만드는 방법은
# 지금 데이터와 같이 종속변수에서는 사용해선 안된다.
# 즉 예측해야되는 값을 줄여버리면 결과도 줄어버린 값이 나온다.
# invers transform으로 역스케일링할 수 있지만
# 그 과정에서 큰 차이가 발생하여 이상한 값이 되버리는 문제가 발생한다.
# 따라서 종속 변수는 스케일링을 하지 않고 그대로 두고 진행한다.
```

```python

cat_col = ['bedrooms', 'bathrooms', 'stories']
num_col = ['price', 'area']

# 인코딩
oh_enc = OneHotEncoder(sparse_output = False) # 범주
cat_data = oh_enc.fit_transform(data[cat_col])
cat_df = pd.DataFrame(cat_data, columns = oh_enc.get_feature_names_out())

# 스케일링
mm_scaler = MinMaxScaler()  # 수치
num_data = mm_scaler.fit_transform(data[num_col])
num_df = pd.DataFrame(num_data, columns = mm_scaler.feature_names_in_)

# 합치기
data_df = pd.concat([cat_df, num_df], axis = 1)
data_df.head()
```
