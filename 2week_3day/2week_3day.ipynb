```python
url = "http://ggoreb.com/python/html/static_practice.html"
res = requests.get(url)
res.encoding = 'utf-8'
soup = bs(res.text, "html.parser")
# 1. 첫번째 도서 제목 추출
a_title = soup.select_one(".book-list .title")
print("도서 제목:", a_title.text) # 혼자 공부하는 파이썬
# 2. 첫번째 li 태그의 data-isbn 속성 추출
li_item = soup.select_one(".item")
isbn = li_item.get("data-isbn")
print("ISBN:", isbn)
# 3. 모든 도서의 가격 합계
price_list = soup.select(".book-list .price")
total_price = 0
for price_tag in price_list:
price_text = price_tag.text
clean_price = price_text.replace(",", "").replace("원", "")
total_price += int(clean_price)
print(f"모든 도서의 가격 합계는 {total_price:,}원입니다.")
```

```python
import pymysql # 모듈 import

conn = pymysql.connect(
    host='svc.sel3.cloudtype.app', user='root', password='1234',
    db='python', charset='utf8', port=31776
) # 데이터베이스 접속

cursor = conn.cursor() # 커서 객체 생성

sql = '''
select * from bang
'''    #데이터베이스에서 요구하는(쓰는) 명령어

cursor.execute(sql) # SQL 실행

result = cursor.fetchall()
for row in result:
    print(row)

conn.commit() # 실행내용 저장
cursor.close() # 커서 객체 종료
conn.close() # 접속 해제
```

```python
import pymysql
from pymysql.cursors import DictCursor

conn = pymysql.connect(
host='127.0.0.1', user='root', password='1234',
db='pythondb', charset='utf8'
)

cursor = conn.cursor(DictCursor)

sql = '''
SELECT id, a, b, c bang
'''
cursor.execute(sql)

result = cursor.fetchall()
for row in result:
    print(row['id'], row['a'], row['b'], row['c'])

cursor.close()
conn.close()
```
