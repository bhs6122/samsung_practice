```python
# 평균을 구하는 함수를 만듬
def my_mean(values):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    return sum / n
```

```python
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)

# agg()를 통해 내가 만든 함수를 적용시킬 수 있다.
```

```python
# 인자 2개를 활용할 수 있다.
def my_mean_diff(values, diff_value):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    mean = sum / n
    return mean - diff_value
```

```python
global_mean = df['lifeExp'].mean()
print(global_mean)
```

```python
agg_mean_diff = df.groupby('year')['lifeExp'].agg(my_mean_diff, diff_value=global_mean)
agg_mean_diff
```
