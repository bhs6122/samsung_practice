```python
# 평균을 구하는 함수를 만듬
def my_mean(values):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    return sum / n
```

```python
agg_my_mean = df.groupby('year')['lifeExp'].agg(my_mean)
print(agg_my_mean)

# agg()를 통해 내가 만든 함수를 적용시킬 수 있다.
```

```python
# 인자 2개를 활용할 수 있다.
def my_mean_diff(values, diff_value):
    n = len(values)
    sum = 0
    for value in values:
        sum += value
    mean = sum / n
    return mean - diff_value
```

```python
global_mean = df['lifeExp'].mean()
print(global_mean)
```

```python
agg_mean_diff = df.groupby('year')['lifeExp'].agg(my_mean_diff, diff_value=global_mean)
agg_mean_diff
```

```python
df.groupby('year')['lifeExp'].mean() - df['lifeExp'].mean()

# 이렇게 해도 위와 같은 결과가 나온다.
# agg기능을 설명하기 위해 위처럼 했지만, 효율적인 방법 골라서 하면 됨
```

#### 다양한 그래프

```python
# 데이터 준비 단계
tips = sns.load_dataset('tips')
tips
```

```python
# 히스토그램: 숫자형 데이터의 분포와 빈도를 표현
# 막대그래프는 막대가 서로 떨어져있지만 히스토그램은 붙어있는게 특징
# 구간별 빈도를 시각화 한 것
hist = plt.hist(tips['total_bill'], bins=10)
plt.title('Histogram of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
```

```python
# 데이터 준비 단계
series = []
for group in tips.groupby('sex')['total_bill']:
    series.append(group[1])
series
```

```python
# 박스 그래프: 집단 간의 분포 차이를 표현
box1 = plt.boxplot(series, tick_labels=['Female', 'Male'])
plt.title('Boxplot of Tips by Gender')
plt.xlabel('Gender')
plt.ylabel('Tip')

# 주황 선이 2분위수. 즉 중앙값(Median)
```

```python
# vert를 통해 가로로 그리기 가능
box2 = plt.boxplot(series, tick_labels=['Female', 'Male'], vert=False)
plt.title('Boxplot of Tips by Gender')
plt.xlabel('Gender')
plt.ylabel('Tip')
```

```python
# 막대 그래프: 집단 간의 차이를 표현할 때 주로 사용
s_tips = tips['sex'].value_counts()
plt.bar(s_tips.index, s_tips.values, color = ['b', 'r'], width = .5)
plt.title('Barplot of Total Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Total Bill Count')
```

```python
s_tips = tips['sex'].value_counts()
plt.pie(s_tips.values, labels = s_tips.index, autopct = '%.2f%%')
plt.title('Pie chart of Total Bill by Sex')
plt.xlabel('Sex')
plt.ylabel('Total Bill Ratio')
```

```python
# 산점도: 데이터의 분포를 표현
plt.scatter(tips['total_bill'], tips['tip'])
plt.title('Scatterplot of Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
```
