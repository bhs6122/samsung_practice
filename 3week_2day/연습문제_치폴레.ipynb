### Step 5. item_price에서 $를 없애고 float 형식으로 새롭게 저장해주세요.

```python
chipo['item_price']     # $가 숫자 맨 앞에 표시된 것을 확인할 수 있음
```

```python
def price(x):               # 맨 앞의 달러 기호를 제거하는 함수
    return float(x[1:]) 

chipo['item_price'] = chipo['item_price'].apply(price)    
# 내가 만든 함수 적용 및 원래 테이블에 변경사항 적용(저장)
```

### Step 6. 아이템별 주문 수를 출력해주세요

```python
# ~별 이런문제는 groupby라고 보면 됨
# 즉 아이템 이름끼리 묶어라는 의미

chipo.groupby('item_name')['order_id'].count()

# order_id에 누락값이 없는걸 알기 때문에 count() 사용
```

### Step 7. 아이템별 주문 총량을 출력해주세요.

```python
chipo.groupby('item_name')['quantity'].sum()
```

### Step 8. 각 아이템의 가격을 구해서 unit_price 컬럼에 추가해주세요.
주문당 quantity가 1일때의 가격 (item_price / quantity)

```python
chipo['unit_price'] = chipo['item_price'] / chipo['quantity']
chipo.head()
```

### Step 9. 아이템별 개당 가격의 평균을 구하고 가장 비싼순으로 5개를 출력해주세요

```python
result = chipo.groupby('item_name')['unit_price'].mean()
result
```

```python
result.sort_values(ascending = False).head()
```

```python
chipo.groupby('item_name')['unit_price'].mean().sort_values(ascending = False).head()
# 한줄로 쓸 수도 있긴 한데 굳이 싶
```

### Step 10. 한 주문(같은 order_id)에 20달러 이상 지불한 손님을 출력해주세요.

```python
chipo_price = chipo.groupby('order_id')['item_price'].sum()

chipo_price[chipo_price >= 20]
```

```python
chipo_price_index = chipo_price[chipo_price >= 20].index
    # 조건에 맞는 데이터의 인덱스만 뽑아올 수 있다.
chipo_price_index
```

### Step 13. Chicken이 포함된 음식은 몇 가지가 있나요?

```python
import re

def check_chicken(x):       # Chicken이라는 단어를 검색하는 함수를 만듬
    return bool(re.search('Chicken', x))    # Chicken단어가 있으면 True
```

```python
chipo['item_name'].apply(check_chicken)
```

```python
chipo[chipo['item_name'].apply(check_chicken)]['item_name'].unique()

# chicken이라는 단어가 들어간 메뉴 종류가 6가지임을 확인할 수 있다.
```

```python
chipo['item_name'].str.contains('Chicken')
# 정규표현식을 쓰지 않고도 Chicken 단어가 있는 데이터를 찾을 수 있다.
```

```python
chipo[chipo['item_name'].str.contains('Chicken')]['item_name'].unique()
```
